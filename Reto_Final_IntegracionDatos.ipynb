#Variables escogidas



runtime, release_year, seasons, imdb_score, imdb_votes, tmdb_popularity, tmdb_score

```python
np.random.seed(42)

XDo = np.array(Netflix[['runtime', 'release_year', 'seasons', 'imdb_score', 'imdb_votes', 'tmdb_popularity', 'tmdb_score']]) #Variables de numero para las bases de datos

XDe1 = np.array(Amazon[['runtime', 'release_year', 'seasons', 'imdb_score', 'imdb_votes', 'tmdb_popularity', 'tmdb_score']]) #Variables de numero para las bases de datos

XDe2 = np.array(Disney[['runtime', 'release_year', 'seasons', 'imdb_score', 'imdb_votes', 'tmdb_popularity', 'tmdb_score']]) #Variables de numero para las bases de datos

XDe3 = np.array(HBO[['runtime', 'release_year', 'seasons', 'imdb_score', 'imdb_votes', 'tmdb_popularity', 'tmdb_score']]) #Variables de numero para las bases de datos
```

```python
#1 Buscamos los máximos y los minimos
XDoM = XDo.max(axis=0)
print("Los valores máximos de las variables son:\n",XDoM)

XDom = XDo.min(axis=0)
print("Los valores mínimos de las variables son:\n",XDom)

#2 ¿Cuales son los 5- valores que me entregará cada variable?
XC=np.zeros((5,7)) # 5 semillas, 7 variables

for j in range(5):
  for i in range(7):
    XC[j,i]=np.random.uniform(XDom[i],XDoM[i])

print("Las semillas (antes):\n",XC)

for k in range (len(XDo)):
  d=np.sqrt(((XC-XDo[k,])**2).sum(axis=1))
  fila=np.where(d==d.min())[0][0]
  XC[fila]=(XC[fila]+XDo[k,])/2

#print("El dato a integrar es:",XDo[0,])
#print("El grupo al que pertenece el primer dato es:",fila)
print("Los clusters (después):\n",XC)

#3. Se procede a obtener los límites de los clusters
sigma=np.zeros((5,1))

for j in range(5):
  sigma[j,]=np.sum(np.sqrt(((XC-XC[j,])**2).sum(axis=1)))/7

print("los diametros sigma de los clusters son:\n",sigma)
```

#Integración Amazon

```python
XCAntes = np.copy(XC)
XDoriginal=XDo.copy()
print("Los datos externos son:\n",XDe1)

#1. SE PROCEDE A PEGAR LOS DATOS EXTERNOS A LOS DATOS INTERNOS
print("Los clusters antes son:\n",XC)

XC1 = np.copy(XC)
#Para saber el numero de datos que se integraron
nd=0
for k in range (len(XDe1)):
  VP=np.exp(-0.5*(((XC1-XDe1[k])/sigma)**2).sum(axis=1))
  fila=np.where(VP==VP.max())[0][0]

  if VP[fila]>0.95:
    nd=nd+1
    XC1[fila]=(XC1[fila]+XDe1[k])/2
    XDo=np.vstack((XDo,XDe1[k]))

#print("los valores de pertenencia son:\n",VP)
#print("El cluster al que pertenece el dato es:",fila)
print("Los datos de la base de datos interna e1 son:", len(XDe1))
print("los datos que se integraron son:",nd)
print("los clusters despues son:\n",XC1)

#2 Como estaba la base de datos antes y la base de datos despues
from scipy.stats import skew #coeficiente de Asimetría

print("Los coeficientes de asimetría antes:\n",skew(XDoriginal,axis=0))
print("Los coeficientes de asimetría despues:\n",skew(XDo,axis=0))
```

```python
#4. se procede con la construcción de los graficos de arañas
#primero definimos los angulos del grafico de araña, un angulo por cada variable.
angles=np.linspace(0, 2 * np.pi, 7, endpoint=False)
angles=np.concatenate((angles, [angles[0]])) #esto es para cerra el circulo de los angulos.
print(angles)               #aqui lo veremos en radianes los cuales s¿van entre 0 y 2*pi
print (angles*100/np.pi) #Esto es para ver los angulos en grados
labels=['runtime', 'release_year', 'seasons', 'imdb_score', 'imdb_votes', 'tmdb_popularity', 'tmdb_score']

#datos antes
XCAntes=XCAntes/np.max(XCAntes,axis=0) #es para normalizar las variables
XCAntes=np.transpose(XCAntes) #para que queden primero las variables y luego los clusters
values= np.concatenate((XCAntes, [XCAntes[0]])) #esto es para cerra el circulo de los angulos.
print(values)

#datos despues
XC1=XC1/np.max(XC1,axis=0) #es para normalizar las variables
XC1=np.transpose(XC1) #para que queden primero las variables y luego los
values2= np.concatenate((XC1, [XC1[0]])) #esto es para cerra el circulo de los angulos.
print(values2)

#5. creo el grafcio de araña
plt.figure()
legend=['antes', 'despues']
plt.polar(angles, values[:,0])                           #para el clusters 1  antes
plt.fill(angles, values [:,0], alpha=0.25)
plt.legend(legend [0])

plt.polar(angles, values2[:,0])                          #para el clusters 1  despues
plt.fill(angles, values2 [:,0], alpha=0.25)
plt.legend(legend [1])

# The change is here: Use angles[:-1] to match the number of labels
plt.thetagrids(angles[:-1]*180/np.pi, labels)  # Use angles[:-1] to select the first 7 angles
plt.title('clusters')
plt.legend(legend,loc='best')
plt.show()

print('los clustes antes son:\n', XCAntes[:,0])
print('Los clustes ahora son:\n', XC1[:,0])
```
